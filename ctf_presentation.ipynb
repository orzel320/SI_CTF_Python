| Sytuacja kolizji | Co dzieje się z flagami | Reakcja postaci |
| :--- | :--- | :--- |
| **1. Żadna postać nie niesie flagi** | Brak zmian - flagi pozostają w bazach | Postacie swobodnie przez siebie przenikają |
| **2. Jedna postać niesie flagę** | Przenoszona flaga natychmiast wraca do swojej bazy | Niosący zostaje odrzucony w kierunku przeciwnym do wroga i ogłuszony na 2,5 s |
| **3. Obie postacie niosą flagę** | Obie flagi wracają do swoich baz | Obie postacie zostają odrzucone od siebie i ogłuszone |

## Funkcja użyteczności — model matematyczny

::: {.math-box}
$$c^* = \arg\max_{c \in C} \Big( U_{\text{base}}(c) + U_{\text{tactical}}(c \mid S) \Big), \qquad U_{\text{base}}(c) = w_{\text{terr}} \cdot T(c) - w_{\text{travel}} \cdot \hat{d}(p_{\text{my}}, c)$$
:::

| Stan flag $(f_{\text{my}}, f_{\text{en}})$ | Tryb taktyczny | Postać składnika $U_{\text{tactical}}(c \mid S)$ |
| :--- | :--- | :--- |
| $(0, 0)$ | **Atak** | $-w_{\text{goal}} \cdot \hat{d}(c, b_{\text{en}})$ <br> *Dążenie po flagę przeciwnika* |
| $(1, 0)$ | **Powrót** | $-w_{\text{goal}} \cdot \hat{d}(c, b_{\text{my}}) - w_{\text{threat}} \cdot \alpha \cdot \Phi(c)$ <br> *Powrót do bazy z barierą odpychającą od wroga ($\epsilon$)* |
| $(0, 1)$ | **Obrona** | Jeśli $\hat{d}(p_{\text{my}}, p_{\text{en}}) \le \beta \hat{d}(p_{\text{en}}, b_{\text{en}})$: $-w_{\text{int}}(\gamma \cdot \text{pościg} + (1-\gamma) \cdot \text{odcięcie})$ <br> *W przeciwnym razie: kontratak* |
| $(1, 1)$ | **Standoff** | Bliżej bazy: powrót; dalej: $-w_{\text{int}} \cdot \hat{d}(c, p_{\text{en}})$ <br> *Wymuszenie resetu* |

: {tbl-colwidths="[18, 16, 66]"}

::: {.footer-note-light}
**Zbiór kandydatów $C$:** Punkty mapy $M$ (z odfiltrowaniem osiągniętych punktów pośrednich w promieniu $r_{\text{arr}}$) oraz pozycja przeciwnika $p_{\text{en}}$, gdy niesie on flagę.
:::

## Optymalizacja parametrów (Optuna) — przebieg treningu

````python
#| echo: false
#| fig-align: center
#| fig-width: 11.4
#| fig-height: 4.5

import math
from pathlib import Path
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from matplotlib.lines import Line2D
from matplotlib.patches import Patch

CSV_PATH = Path("params.csv")
ROLLING_WINDOW = None
SHOW_RAW_POINTS = True

DEFAULT_PARAMS = {
    "w_goal": 2.0,
    "w_travel": 0.5,
    "w_threat": 3.0,
    "w_int": 2.5,
    "w_terr": 0.4,
    "alpha": 0.15,
    "beta": 1.15,
    "gamma": 0.6,
    "epsilon": 0.1,
    "r_arr": 50.0,
}

PARAM_LABELS = {
    "w_goal": "w_goal (cel bazy)",
    "w_travel": "w_travel (koszt ruchu)",
    "w_threat": "w_threat (unikanie wroga)",
    "w_int": "w_int (przechwytywanie)",
    "w_terr": "w_terr (przewaga teryt.)",
    "alpha": "alpha (skala zagrożenia)",
    "beta": "beta (próg pościgu)",
    "gamma": "gamma (pościg / odcięcie)",
    "epsilon": "epsilon (wygładzanie)",
    "r_arr": "r_arr (promień celu)",
}

PALETTE = ["#2563EB", "#D97706", "#DC2626", "#059669", "#7C3AED", "#0891B2", "#BE185D", "#4F46E5", "#CA8A04", "#0D9488"]

df = pd.read_csv(CSV_PATH).sort_values("occurrence").reset_index(drop=True)

param_cols = [c for c in DEFAULT_PARAMS if c in df.columns]
varying_cols = [c for c in param_cols if df[c].std(ddof=0) > 1e-9]
fixed_cols = [c for c in param_cols if c not in varying_cols]

if not varying_cols:
    varying_cols = param_cols[:4]
    fixed_cols = param_cols[4:]

n_iter = len(df)
win = ROLLING_WINDOW if ROLLING_WINDOW is not None else max(5, min(15, n_iter // 10))

plt.rcParams.update({
    "figure.facecolor": "#F8FAFC",
    "axes.facecolor": "#FFFFFF",
    "axes.edgecolor": "#CBD5E1",
    "axes.labelcolor": "#334155",
    "xtick.color": "#475569",
    "ytick.color": "#475569",
    "text.color": "#0F172A",
    "font.sans-serif": ["Inter", "DejaVu Sans"],
})

n_plots = len(varying_cols)
if n_plots <= 3:
    nrows, ncols = 1, n_plots
elif n_plots == 4:
    nrows, ncols = 2, 2
elif n_plots <= 6:
    nrows, ncols = 2, 3
else:
    nrows, ncols = 2, math.ceil(n_plots / 2)

fig, axes = plt.subplots(nrows, ncols, figsize=(11.4, 4.5), dpi=160, sharex=True)
axes_flat = np.atleast_1d(axes).ravel()

for idx, col in enumerate(varying_cols):
    ax = axes_flat[idx]
    color = PALETTE[idx % len(PALETTE)]
    x = df["occurrence"]
    y = df[col]

    roll = y.rolling(window=win, min_periods=1, center=True)
    med = roll.median()
    q25 = roll.quantile(0.25)
    q75 = roll.quantile(0.75)

    ax.fill_between(x, q25, q75, color=color, alpha=0.14, zorder=1)
    if SHOW_RAW_POINTS:
        ax.scatter(x, y, color=color, alpha=0.28, s=15, edgecolors="none", zorder=2)
    ax.plot(x, med, color=color, linewidth=2.2, zorder=4)

    default_val = DEFAULT_PARAMS[col]
    ax.axhline(default_val, color="#64748B", linestyle="--", linewidth=1.15, alpha=0.8, zorder=3)

    # Zwięzły opis w narożniku panelu: wartość początkowa -> mediana końcówki treningu
    tail_med = y.iloc[-win:].median()
    ax.set_title(PARAM_LABELS.get(col, col), fontsize=9.8, fontweight="bold", loc="left", pad=5)
    ax.text(
        0.99, 1.03,
        f"domyślna: {default_val:g}  |  koniec (med.): {tail_med:.2f}",
        transform=ax.transAxes,
        ha="right", va="bottom",
        fontsize=7.6, color="#475569",
    )

    ax.grid(True, linestyle=":", color="#CBD5E1", alpha=0.8)
    ax.spines["top"].set_visible(False)
    ax.spines["right"].set_visible(False)
    ax.tick_params(axis="both", labelsize=8)

for ax in axes_flat[n_plots:]:
    ax.set_visible(False)

for ax in axes_flat[max(0, n_plots - ncols):n_plots]:
    ax.set_xlabel("Iteracja treningu (occurrence)", fontsize=8.5, labelpad=4)

legend_handles = [
    Line2D([0], [0], color="#334155", lw=2.2, label=f"Mediana krocząca (okno = {win})"),
    Patch(facecolor="#334155", alpha=0.18, label="Rozstęp IQR (25%–75%)"),
    Line2D([0], [0], color="#64748B", lw=1.2, ls="--", label="Wartość domyślna (DEFAULT_PARAMS)"),
]
if SHOW_RAW_POINTS:
    legend_handles.insert(
        0,
        Line2D([0], [0], marker="o", color="w", markerfacecolor="#334155", alpha=0.45, markersize=5.5, label="Pojedyncza próba Optuny"),
    )

fig.legend(
    handles=legend_handles,
    loc="upper center",
    bbox_to_anchor=(0.5, 0.935),
    ncol=len(legend_handles),
    frameon=False,
    fontsize=8.2,
    columnspacing=1.5,
)

total_min = df["elapsed_sec"].max() / 60.0 if "elapsed_sec" in df.columns else 0.0
fixed_summary = ", ".join(f"{c}={df[c].iloc[0]:g}" for c in fixed_cols)
title_parts = [f"Iteracje: {n_iter}", f"Łączny czas: {total_min:.1f} min"]
if fixed_summary:
    title_parts.append(f"Parametry stałe: {fixed_summary}")

fig.suptitle("  |  ".join(title_parts), fontsize=8.5, color="#475569", y=0.99)
fig.subplots_adjust(left=0.055, right=0.985, bottom=0.11, top=0.81, wspace=0.18, hspace=0.36)
plt.show()
````
